```python
import random
import numpy as np
import torch

# ==========================================
# BÀI 1: Tạo Tensor cơ bản
# ==========================================
a1 = torch.tensor([[1-3], [4-6]], dtype=torch.int32)
print("=== BÀI 1 ===")
print("Tensor a:\n", a1)
print("Shape / Size:", a1.size())
print("Số chiều (ndim):", a1.ndim)
print("Kiểu dữ liệu (dtype):", a1.dtype)

# ==========================================
# BÀI 2: Tạo các Tensor thường dùng
# ==========================================
a_rand = torch.rand(3, 4)
b_randn = torch.randn(3, 4)
c_zeros = torch.zeros(2, 3)
d_ones = torch.ones(2, 3)
e_eye = torch.eye(4)
tensor_10 = torch.full((5, 5), 10)

print("\n=== BÀI 2 ===")
print("rand (Uniform [0, 1)):\n", a_rand)
print("randn (Normal mean=0, std=1):\n", b_randn)
print("Tensor 5x5 toàn giá trị 10:\n", tensor_10)

# ==========================================
# BÀI 3: Kiểm tra thông tin Tensor
# ==========================================
a_3d = torch.zeros(2, 3, 4)
print("\n=== BÀI 3 ===")
print("Shape:", a_3d.shape)
print("Size:", a_3d.size())
print("Dimensions:", a_3d.ndim)
print("Data type:", a_3d.dtype)

# ==========================================
# BÀI 4: Slicing và Indexing
# ==========================================
a_4 = torch.randn(3, 4, 5)
print("\n=== BÀI 4 ===")
print("Shape a:", a_4.shape)
print("Shape a[1:, 2:4]:", a_4[1:, 2:4].shape)
print("Shape 2 phần tử cuối chiều 3:", a_4[:, :, -2:].shape)

# ==========================================
# BÀI 5: Reshape và Flatten
# ==========================================
a_5 = torch.randn(3, 4)
print("\n=== BÀI 5 ===")
print("Số phần tử ban đầu:", a_5.numel())
print("Số phần tử sau reshape (3x2x2):", a_5.reshape(3, 2, 2).numel())
print("Tensor reshape 2x6:\n", a_5.reshape(2, 6))

# ==========================================
# BÀI 6: Phép toán và nhân ma trận
# ==========================================
a_6, b_6 = torch.randn(2, 3), torch.randn(2, 3)
print("\n=== BÀI 6 ===")
print("a * b (element-wise):\n", a_6 * b_6)
print("a @ b.T (matrix multiplication):\n", a_6 @ b_6.T)
print("Shape của a @ b.T:", (a_6 @ b_6.T).shape)

# ==========================================
# BÀI 7: Các hàm thống kê trên Tensor
# ==========================================
a_7 = torch.randn(3, 4)
print("\n=== BÀI 7 ===")
print("dim=0 (theo cột):\n", torch.mean(a_7, dim=0))
print("dim=1 (theo hàng):\n", torch.mean(a_7, dim=1))

# ==========================================
# BÀI 8: Autograd cơ bản
# ==========================================
x_8a = torch.tensor(3.6, requires_grad=True)
(x_8a ** 2).backward()
print("\n=== BÀI 8 ===")
print("dy/dx (y=x^2) tại x=3.6:", x_8a.grad.item())

x_8b = torch.tensor(3.6, requires_grad=True)
(3 * x_8b**2 + 2 * x_8b + 1).backward()
print("dy/dx (y=3x^2+2x+1) tại x=3.6:", x_8b.grad.item())

# ==========================================
# BÀI 9: Gradient Descent tìm hệ số đa thức
# ==========================================
poly9 = np.poly1d([1-3])
X9 = np.random.randn(20, 1) * 5
Y9 = poly9(X9)
XX9 = np.hstack([X9 * X9, X9, np.ones_like(X9)])
w9 = torch.randn(3, 1, requires_grad=True)
opt9 = torch.optim.NAdam([w9], lr=0.01)

for _ in range(1000):
    opt9.zero_grad()
    loss9 = torch.mean((torch.tensor(XX9, dtype=torch.float32) @ w9 - torch.tensor(Y9, dtype=torch.float32))**2)
    loss9.backward()
    opt9.step()

print("\n=== BÀI 9 ===")
print("Hệ số học được:", w9.detach().numpy().flatten())

# ==========================================
# BÀI 10: Giải hệ phương trình bằng Autograd
# ==========================================
A_10 = torch.tensor(random.random(), requires_grad=True)
B_10 = torch.tensor(random.random(), requires_grad=True)
C_10 = torch.tensor(random.random(), requires_grad=True)
D_10 = torch.tensor(random.random(), requires_grad=True)
opt10 = torch.optim.NAdam([A_10, B_10, C_10, D_10], lr=0.01)

for _ in range(2000):
    opt10.zero_grad()
    loss10 = (A_10+B_10-9)**2 + (C_10-D_10-1)**2 + (A_10+C_10-8)**2 + (B_10-D_10-2)**2
    loss10.backward()
    opt10.step()

print("\n=== BÀI 10 ===")
print(f"A={A_10.item():.2f}, B={B_10.item():.2f}, C={C_10.item():.2f}, D={D_10.item():.2f}")

# ==========================================
# BÀI 11: Bài tập áp dụng tổng hợp
# ==========================================
print("\n=== BÀI 11 ===")
X11 = torch.arange(24)
print("1. Reshape 4x6:\n", X11.reshape(4, 6))

A11, B11 = torch.randn(3, 4), torch.randn(3, 4)
print("2. A + B:\n", A11 + B11)
print("   Mean cột A:", torch.mean(A11, dim=0))

x11c = torch.tensor(2.0, requires_grad=True)
(2 * x11c**2 + 5 * x11c + 3).backward()
print("3. dy/dx tại x=2:", x11c.grad.item())

poly11d = np.poly1d([2, -3, 5])
X11d = np.random.randn(20, 1) * 5
Y11d = poly11d(X11d)
XX11d = np.hstack([X11d * X11d, X11d, np.ones_like(X11d)])
w11d = torch.randn(3, 1, requires_grad=True)
opt11d = torch.optim.NAdam([w11d], lr=0.01)

for _ in range(1000):
    opt11d.zero_grad()
    loss11 = torch.mean((torch.tensor(XX11d, dtype=torch.float32) @ w11d - torch.tensor(Y11d, dtype=torch.float32))**2)
    loss11.backward()
    opt11d.step()

print("4. Hệ số học được:", w11d.detach().numpy().flatten())
```
